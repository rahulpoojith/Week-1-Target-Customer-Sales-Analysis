# Amazon Dataset Cleaning

This notebook is for user's reference to clean  and save the data before doing data analysis.

## Step 1: Import pandas

Pandas helps us read and edit the table.

```python
import pandas as pd
```

## Step 2: Read the raw data

The raw CSV file is in the `Datasets` folder. `head()` shows a few rows so we can see the data. 

```python
df = pd.read_csv("../Datasets/amazon.csv")
df.head()
```

## Step 3: Clean text spacing

Remove extra spaces around text values. Turn the `|` placeholder in ratings into a missing value so it can be handled correctly.

```python
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()

df["rating"] = df["rating"].replace("|", pd.NA)
```

```python
print("Rows before removing duplicates:", len(df))
df.count()
```

## Step 4: Remove duplicate products

Keep one row for each product ID.

```python
df = df.drop_duplicates(subset="product_id")
```

```python
print("Rows after removing duplicates:", len(df))
df.count()
```

```python
df.head(5)
```

## Step 5: Clean price and percentage columns

Remove currency symbols, commas, and percent signs, then convert these columns to numbers.

```python
df["discounted_price"] = pd.to_numeric(
    df["discounted_price"].astype(str).str.replace("[₹,]", "", regex=True),
    errors="coerce"
)

df["actual_price"] = pd.to_numeric(
    df["actual_price"].astype(str).str.replace("[₹,]", "", regex=True),
    errors="coerce"
)

df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)
```

```python
df.head(5)
```

## Step 6: Clean rating columns

Convert ratings and rating counts to numbers. Values that cannot be converted become missing values.

```python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

df["rating_count"] = pd.to_numeric(
    df["rating_count"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
df.head(5)
```

## Step 7: Remove rows missing key values

Remove rows that do not have a discounted price, rating, or rating count.

```python
df = df.dropna(subset=["discounted_price", "rating", "rating_count"])
df.head()
```

## Step 8: Add a main category

The category column contains several levels separated by `|`. Keep the first level in a new column.

```python
df["main_category"] = df["category"].str.split("|").str[0]
df[["category", "main_category"]].head()
```

## Step 9: Remove columns not needed for product analysis

Keep product names, categories, prices, discounts, ratings, and rating counts. Remove reviewer IDs and names, review IDs and text, and image and product links. The reviewer fields describe people who left reviews; they do not show recorded purchases, so they cannot tell us what customers bought.

This dataset contains product listings and reviews, not order records or units sold.

```python
df = df.drop(columns=[
    "user_id",
    "user_name",
    "review_id",
    "review_title",
    "review_content",
    "img_link",
    "product_link"
])
df.head()
```

## Step 10: Save the cleaned data

Save the cleaned table as a new CSV file in `Processed_Datasets`.

```python
df.to_csv("../Processed_Datasets/amazon_cleaned.csv", index=False)
print("Cleaned data saved.")

print("Rows in cleaned data:", len(df))
df.head()
```
